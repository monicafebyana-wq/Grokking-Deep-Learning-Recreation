## 1. Apa itu Deep Learning?

Deep learning merupakan bagian dari machine learning dan artificial intelligence.

Menurut buku, deep learning adalah bidang yang berfokus pada otomatisasi kecerdasan menggunakan neural network.

Penerapannya banyak digunakan pada:
- Computer vision
- Speech recognition
- Machine translation
- Berbagai masalah industri

Deep learning menjadi penting karena mampu mempelajari pola kompleks dari data.

## 2. Mengapa Belajar Deep Learning?

Alasan utama mempelajari deep learning:

1. **Powerful tool untuk otomatisasi kecerdasan**

   Deep learning membantu komputer melakukan tugas yang sebelumnya membutuhkan kemampuan manusia.

2. **Banyak digunakan pada industri modern**

   Contoh:
   - Analisis gambar
   - Pengenalan suara
   - Sistem rekomendasi

3. **Bidang yang kreatif**

   Membangun sistem yang dapat belajar memberikan pemahaman tentang bagaimana kecerdasan dapat dimodelkan.

## 6. Menggunakan NumPy Array

Deep learning banyak menggunakan operasi array.

NumPy membantu melakukan perhitungan numerik dengan lebih mudah.

```python
angka = np.array([1,2,3,4,5])

print(angka)
print("Rata-rata:", np.mean(angka))
print("Jumlah:", np.sum(angka))
```

## 7. Latihan Mini: Simulasi Prediksi Sederhana

Sebelum neural network, kita membuat model sederhana.

Model menerima input dan mengalikan dengan sebuah nilai (weight).

Konsep weight akan dipelajari lebih dalam pada chapter berikutnya.

```python
input_data = 5
weight = 0.5

prediction = input_data * weight

print("Input:", input_data)
print("Weight:", weight)
print("Prediction:", prediction)
```
