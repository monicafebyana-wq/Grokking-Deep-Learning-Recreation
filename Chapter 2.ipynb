# 2. Apa itu Deep Learning?

Deep Learning adalah kumpulan metode dalam machine learning yang menggunakan artificial neural network.

Contoh penggunaan:
- Computer Vision (gambar)
- Natural Language Processing (teks)
- Speech Recognition (suara)

Tujuan utamanya adalah membuat komputer mampu menemukan pola dari data.

# 3. Apa itu Machine Learning?

Machine Learning adalah bidang yang membuat komputer dapat belajar melakukan tugas tanpa diprogram secara eksplisit.

Ide sederhananya:

Data → Machine Learning Model → Prediksi

Model mengamati pola dari data kemudian menggunakannya untuk membuat prediksi baru.

```python
# Contoh sederhana machine learning:
# Menggunakan data lama untuk membuat prediksi

data_training = [
    (1, 2),
    (2, 4),
    (3, 6)
]

print(data_training)
```

# 4. Predict, Compare, Learn

Buku memperkenalkan pola pembelajaran utama:

1. Predict
   - Model menghasilkan prediksi berdasarkan input.

2. Compare
   - Prediksi dibandingkan dengan hasil sebenarnya.

3. Learn
   - Model memperbaiki parameter agar prediksi berikutnya lebih baik.

Siklus ini menjadi dasar pembelajaran neural network.

```python
# Simulasi predict, compare, learn sederhana

weight = 0.5
input_data = 10
target = 12

prediction = input_data * weight

error = target - prediction

print("Prediction:", prediction)
print("Error:", error)

# proses belajar sederhana
weight = weight + error * 0.01

print("Updated weight:", weight)
```

# 5. Supervised Machine Learning

Supervised learning mempelajari hubungan antara input dan output yang sudah diketahui.

Contoh:
- Gambar → apakah ada kucing?
- Data cuaca → kemungkinan hujan
- Data film → rekomendasi film

Dalam supervised learning terdapat:
Input Dataset → Model → Output Dataset

```python
# Contoh data supervised learning sederhana

input_data = [
    "panas",
    "dingin",
    "hangat"
]

output_data = [
    "minum",
    "pakai jaket",
    "normal"
]

for x,y in zip(input_data, output_data):
    print(x, "->", y)
```

# 6. Unsupervised Machine Learning

Unsupervised learning mencari pola dari data tanpa memiliki jawaban benar sebelumnya.

Tujuan utama:
- menemukan kelompok
- menemukan struktur tersembunyi

Contoh:
Mengelompokkan pelanggan berdasarkan kebiasaan belanja.

```python
# Contoh sederhana clustering manual

data = [
    [1,2],
    [2,3],
    [10,11],
    [11,12]
]

print("Data:")
for d in data:
    print(d)
```

# 7. Parametric dan Nonparametric Learning

## Parametric Learning

Model memiliki jumlah parameter tertentu.

Pembelajaran dilakukan dengan mengubah parameter tersebut.

Contoh:
- Neural Network
- Linear Regression


## Nonparametric Learning

Jumlah parameter mengikuti data yang tersedia.

Model dapat bertambah kompleks sesuai jumlah data.

```python
# Contoh parameter pada model sederhana

weight = 0.5
bias = 1

input_data = 4

output = weight * input_data + bias

print(output)
```

# 8. Empat Jenis Machine Learning

Machine learning dapat dibagi menjadi:

1. Supervised Parametric
2. Supervised Nonparametric
3. Unsupervised Parametric
4. Unsupervised Nonparametric

Deep learning termasuk supervised maupun unsupervised, dan menggunakan model parametric.
