# 1. Konsep Utama: Neural Network Membuat Prediksi

Pada chapter sebelumnya kita mengenal konsep:

**Predict → Compare → Learn**

Chapter 3 fokus pada tahap pertama yaitu **Predict**.

Neural network menerima data input, memprosesnya melalui weight, kemudian menghasilkan output atau prediksi.

Secara sederhana:

Input → Weight → Prediction

Weight adalah nilai yang menentukan seberapa besar pengaruh input terhadap output.

# 2. Neural Network Paling Sederhana

Model pertama hanya memiliki:
- satu input
- satu weight
- satu output

Rumus:

prediction = input × weight

Walaupun sederhana, konsep ini adalah dasar dari jaringan neural yang lebih kompleks.

```python
# Neural network sederhana

def neural_network(input_value, weight):
    prediction = input_value * weight
    return prediction


input_value = 8
weight = 0.5

prediction = neural_network(input_value, weight)

print("Input:", input_value)
print("Weight:", weight)
print("Prediction:", prediction)
```

# 3. Memahami Weight

Weight dapat dianggap sebagai tingkat kepercayaan model terhadap input.

Contoh:

Jika weight = 1:
- input diteruskan hampir sama

Jika weight = 0.5:
- output menjadi setengah input

Jika weight kecil:
- pengaruh input semakin kecil

```python
input_value = 10

weights = [0.1, 0.5, 1, 2]

for w in weights:
    prediction = input_value * w
    print("Weight:", w, "Prediction:", prediction)
```

# 4. Membuat Prediksi dengan Multiple Inputs

Pada kasus nyata, sebuah prediksi biasanya tidak hanya menggunakan satu informasi.

Contoh:
Prediksi kemenangan pertandingan dapat menggunakan:
- jumlah pemain
- statistik pertandingan
- kondisi lapangan

Setiap input memiliki weight masing-masing.

Rumus:

prediction =
(input1 × weight1) +
(input2 × weight2) +
(input3 × weight3)

```python
# Multiple input neural network

def neural_network_multiple_inputs(inputs, weights):
    prediction = 0

    for i in range(len(inputs)):
        prediction += inputs[i] * weights[i]

    return prediction


inputs = [2, 4, 6]
weights = [0.5, 0.2, 0.1]

prediction = neural_network_multiple_inputs(inputs, weights)

print("Prediction:", prediction)
```

# 5. Menggunakan Dot Product dengan NumPy

Dalam deep learning, operasi perkalian input dan weight sangat sering dilakukan.

NumPy menyediakan dot product untuk melakukan operasi tersebut secara efisien.

Rumus:

prediction = input · weight

```python
import numpy as np

inputs = np.array([2,4,6])
weights = np.array([0.5,0.2,0.1])

prediction = np.dot(inputs, weights)

print("Prediction:", prediction)
```

# 6. Multiple Inputs dan Multiple Outputs

Neural network juga dapat menghasilkan lebih dari satu output.

Contoh:
Sebuah model dapat memprediksi:
- kemungkinan menang
- kemungkinan kalah

Setiap output memiliki kumpulan weight sendiri.

```python
inputs = np.array([2,4,6])

weights_output1 = np.array([0.5,0.2,0.1])
weights_output2 = np.array([0.1,0.3,0.5])


prediction1 = np.dot(inputs, weights_output1)
prediction2 = np.dot(inputs, weights_output2)


print("Output 1:", prediction1)
print("Output 2:", prediction2)
```

# 7. Predicting on Predictions

Salah satu konsep penting neural network adalah output dari satu proses dapat menjadi input untuk proses berikutnya.

Hal ini menjadi dasar dari jaringan bertingkat (*deep neural network*).

Contoh:

Input → Layer 1 → Layer 2 → Output

```python
# Layer pertama

input_value = 5

weight1 = 0.5

layer1_output = input_value * weight1


# Layer kedua

weight2 = 2

final_prediction = layer1_output * weight2


print("Layer 1 output:", layer1_output)
print("Final prediction:", final_prediction)
```

# 8. Visualisasi Konsep Forward Propagation

Forward propagation adalah proses mengalirkan informasi dari input menuju output.

Alur:

Input Data
↓
Perkalian dengan Weight
↓
Menghasilkan Prediction

Pada deep neural network, proses ini terjadi pada banyak layer.

# 9. Latihan Mini

Buat fungsi neural network yang menerima:
- input array
- weight array

Kemudian menghasilkan prediksi menggunakan dot product.

```python
def neural_prediction(inputs, weights):
    return np.dot(inputs, weights)


inputs = np.array([3,5,7])
weights = np.array([0.2,0.4,0.6])


result = neural_prediction(inputs, weights)

print(result)
```
