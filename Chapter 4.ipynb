# Chapter 4: Introduction to Neural Learning: Gradient Descent

Chapter 4 membahas bagaimana neural network mulai belajar.

Pada chapter sebelumnya:

Pada chapter ini:
- Neural network belajar memperbaiki prediksi dengan mengurangi error.

Materi:
- Predict, Compare, Learn
- Mengukur error
- Hot and cold learning
- Gradient descent
- Weight update
- Learning rate (alpha)

# 1. Dari Prediksi Menuju Pembelajaran

Sebuah neural network belum disebut belajar jika hanya menghasilkan prediksi.

Agar dapat belajar, neural network harus:

1. Membuat prediksi (**predict**)
2. Membandingkan prediksi dengan nilai sebenarnya (**compare**)
3. Mengubah parameter agar prediksi berikutnya lebih baik (**learn**)

Siklus ini menjadi dasar proses training neural network.

# 2. Predict

Tahap pertama adalah membuat prediksi menggunakan weight.

Rumus sederhana:

prediction = input × weight

Contoh:

```python
def neural_network(input_value, weight):
    return input_value * weight


input_value = 5
weight = 0.5

prediction = neural_network(input_value, weight)

print("Prediction:", prediction)
```

# 3. Compare: Mengukur Error

Setelah mendapatkan prediksi, model harus mengetahui seberapa salah prediksinya.

Error sederhana:

error = prediction - target

Semakin kecil error, semakin baik model.

```python
prediction = 2.5
target = 5

error = prediction - target

print("Prediction:", prediction)
print("Target:", target)
print("Error:", error)
```

# 4. Mengapa Error Harus Diukur?

Tanpa mengetahui error, model tidak tahu apakah weight harus:
- dinaikkan
- diturunkan

Error memberikan informasi tentang kualitas prediksi.

Tujuan learning adalah:

Mengurangi error seminimal mungkin.

# 5. Hot and Cold Learning

Sebelum gradient descent, buku memperkenalkan metode sederhana:

- Mengubah weight sedikit ke arah tertentu
- Melihat apakah error menjadi lebih kecil atau lebih besar

Jika error mengecil:
→ arah perubahan weight benar

Jika error membesar:
→ arah perubahan weight salah

```python
input_value = 5
target = 10

weight = 1

# prediksi awal
prediction = input_value * weight

error = target - prediction

print("Initial prediction:", prediction)
print("Initial error:", error)


# mencoba menaikkan weight
weight = weight + 0.1

prediction_new = input_value * weight

error_new = target - prediction_new

print("New prediction:", prediction_new)
print("New error:", error_new)
```

# 6. Masalah Hot and Cold Learning

Metode hot and cold:
- lambat
- membutuhkan banyak percobaan
- tidak mengetahui seberapa besar perubahan yang harus dilakukan

Karena itu digunakan metode yang lebih efisien: Gradient Descent

# 7. Gradient Descent

Gradient descent adalah metode untuk mencari arah perubahan weight yang paling tepat.

Ide dasarnya:

Jika error besar:
→ ubah weight

Jika error kecil:
→ pertahankan

Tujuan:
mencari nilai weight yang menghasilkan error minimum.

# 8. Learning Rate (Alpha)

Learning rate menentukan besar langkah perubahan weight.

Rumus sederhana:

weight_delta = error × input × alpha

alpha kecil:
- belajar lebih lambat
- lebih stabil

alpha besar:
- belajar lebih cepat
- risiko melewati nilai terbaik

```python
input_value = 5
target = 10

weight = 0.5
alpha = 0.1


prediction = input_value * weight

error = target - prediction

weight_delta = error * input_value * alpha

weight = weight + weight_delta


print("Prediction:", prediction)
print("Error:", error)
print("Weight update:", weight_delta)
print("New weight:", weight)
```

# 9. Implementasi Training Sederhana

Sekarang kita menjalankan beberapa iterasi.

Setiap iterasi:
1. Membuat prediksi
2. Menghitung error
3. Mengupdate weight

```python
input_value = 5
target = 10

weight = 0

alpha = 0.1


for iteration in range(10):

    prediction = input_value * weight

    error = target - prediction

    weight_delta = error * input_value * alpha

    weight += weight_delta

    print(
        "Iteration:", iteration,
        "Prediction:", prediction,
        "Error:", error,
        "Weight:", weight
    )
```

# 10. Gradient Descent dengan Multiple Data

Dalam masalah nyata, neural network belajar dari banyak contoh.

Semakin banyak data:
- model menemukan pola lebih baik
- prediksi menjadi lebih akurat

Proses training dilakukan berulang terhadap seluruh dataset.

```python
inputs = [1,2,3,4]
targets = [2,4,6,8]

weight = 0
alpha = 0.01


for i in range(len(inputs)):

    prediction = inputs[i] * weight

    error = targets[i] - prediction

    weight_delta = error * inputs[i] * alpha

    weight += weight_delta


print("Final weight:", weight)
```
